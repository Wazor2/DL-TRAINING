```python
#Data loading first cell 
from langchain_community.document_loaders import TextLoader
loader = TextLoader('task1data.txt', encoding='utf-8')
cppdata = loader.load()
cppdata
 
```

```python
#Text splitter 

from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=200)
finalDocuments = text_splitter.split_documents(cppdata)
finalDocuments
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
print(embeddings)
```

```python
# Embedding and Vector store in vector_DB
from langchain_community.vectorstores import FAISS
db = FAISS.from_documents(finalDocuments, embeddings)
db 
```

```python
query =input("Enter your query: ")
docs = db.similarity_search(query)
for i in range(len(docs)):
    
    print(docs[i].page_content)
```
